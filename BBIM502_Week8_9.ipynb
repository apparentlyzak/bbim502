```python
def add_cream(amount):
    if amount < 0:
        raise ValueError("Cannot add negative amounts of cream.")
    global cream_amount
    cream_amount += amount
    print("Thanks for the", amount, "cups of cream!")

# The 'global' keyword tells Python that we want to use the global (outside) variable 'cream_amount',
# not create a new local one. This allows the function to modify the global value directly.
#It is necessary when you want to update a value that is shared across multiple functions or parts of the program.
```

```python
def make_ice_cream():
    global cream_amount
    global ice_amount

    if cream_amount <= 0:
        raise RuntimeError("Not enough cream to make ice cream.")
    elif ice_amount <= 0:
        raise RuntimeError("Not enough ice to make ice cream.")
```
